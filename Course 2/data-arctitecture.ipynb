## Why Data Architecture Matters

https://www.youtube.com/watch?v=LnIvDZ-mTnA

Why is it important?

When done well, it can help an organization gain a better understanding of their own data. Not surprisingly, and especially in larger organizations with many teams, data can be represented in many types, many storage locations, multiple original source locations, many sizes, many versions, and used for different purposes. In some cases, there may exist data that is unknown to different teams. Data architecture can provide effective guidelines for managing data from it's initial capture in (multiple?) source systems all the way to the other end to information consumption by business units. That same architecture can additionally provide a tangible and defined structure upon which systems and processes are developed and implemented, which helps to enable effective data governance.

Know your data

Bad data quality can have a tremendous impact on the bottom line for companies, so it matters. That data can lead to incorrect, inconsistent, or even lack of decisions. One of the first steps in data architecture is to document all the details about your data is. Spend the time to really look, so you can be sure what data is ingested into your systems is high quality.

No multiple copies of data

In the same spirit as "know your data", having duplicate data within a system is a recipe for out-of-sync, inaccurate, incorrect, and inconsistent data. At a minimum, it can cause errors and problems during ingestion and storage. If this data is then used in reporting, it can lead to poor decision-making.

## Importance of Data Architecture

https://www.youtube.com/watch?v=-QXy8nONulI

Effective data architecture can enable an organization to clearly see what, where, and how their data resides. This allows your IT teams to ensure data is efficiently secured, optimally stored, and processed accurately. As the organization grows, along with its data, that same data architecture can enable the organization to evaluate the impact, which can then be communicated to all relevant teams. Fata architecture is a (set of) live document(s) that represent the entire collection of systems and data, including the business rules and policies. This living document can guarantee up-to-date data, while also providing a high-level (and low level as needed) end-to-end picture.

The hidden gem of designing data systems

The research involved in design a data system forces people across the organization to talk to each other and collaborate. Those discussions are gold to an organization.

Creating an enterprise data architecture starts with solid requirements and stakeholders to fund the project

What else can be part of an Enterprise Data Architecture?

You may have several layers in your data architecture. At the highest level, you can keep it more abstract so that a wider variety of business stakeholders can understand. Because one of the goals is also to guide implementation, it may add value(with a new layer of detail) to your enterprise data model to include some essential data attributes, data stewardship (ownership) assignments, specific data quality requirements, and appropriate reference data values. You can add as many layers as needed to capture all the information.
