What is a data anomaly?
Inconsistencies/differences in style, type, or formatting in the data stored in a database. These are especially important on the initial data ingestion and during updates. Anomalies must be addressed to avoid inconsistency or incorrect data or reporting. It can involve cleansing and transforming. Identifying anomalies is especially important when ingesting data from multiple sources, data cleaning fixes or removes incorrect, corrupted, incorrectly formatted, duplicate, or incomplete data. It can involve transforming and converting data from other sources into a common readable and compatible format

Bugs in a popular e-commerce system can impact 6,500 orders every minute valued at approximately $300,000. If ODS has a bug in a transformation rule that maps the wrong product code from the stock inventory system to the product code of online orders, as a result, the wrong product is shipped. This transaction triggers a notification to replenish the wrong product.

Anomalies of Nulls	Null values represent different meanings in data. Understanding and handling these is vital when dealing with other data sources
Anomaly	Out of character or unexpected data format, contents, timing, or size
Cleansing	Transforming and converting data from other sources into a common readable and compatible format
Column Mapping	Mapping columns from different datasets to each other
Conversions	Taking one format of data and changing it into another format. For example mm/dd/yy to dd/mm/yyyy
Decoding/encoding	Certain sensitive data such as SSN and date of birth needs to be encoded/decoded in the database
ER model	Technique to related Entities of a business domain
First Normal	Every row in the table has to be unique by creating a primary key column.
Governance	Rules, practices, and processes for management of data assets within an organization
ISO	International Standards Organization
Logical diagram	Visual diagram that represents major sources of data flow and data storage within a system
Normalization	Structuring a relational database in such a way that there is reduced data redundancy and improved data integrity. Uses first 3 forms
ODS	An Operational Data Store is a centralized database that provides a current/recent "snapshot" of the data from all ingested transactional systems. It is especially used for operational reporting. It allows organizations to combine data from the original format, even if different, from various external and internal sources into a single centralized location, which provides access for business reporting.
Reformatting	Data needs to be reformatted when putting in Word or emails
Second Normal	In addition to First normal rule, every non-prime attribute of the relationship is dependent on the whole of every candidate key
Third Normal	In addition to Second normal rule, all non-prime attributes depend only on the candidate keys and do not have a transitive dependency on another key. Why does that matter? Transitive functional dependency means that a non-key column value was determining another non-key column value.
Transformations	Changing the data from one format to another so as to provide a common format
