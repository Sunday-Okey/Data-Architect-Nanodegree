## Deep Dive into History

1960s hierarchical and network models

Some of the first systems that were introduced by IBM, were called the hierarchical database model, which were tree-like structures as shown below. These were based on hierarchies in the real world such as in the military, public library, or schools. The entire tree might need to be traversed in order to access data. The main disadvantage is that it can be challenging to create and show relationships within the data.

Hierarchical database model

In 1969, CODASYL (Conference on Data Systems Languages) published a description of the next-generation model called the network database model, as shown below. Notice any record can have more than one parent and more than 1 child record.

The network database model

These models were used heavily in the 70s but still did not offer true relationships in data. However, also in 1969 the world also learned of another type of database model, called the "relational model." While it did exist, it did not really take hold until the 80s, but the seed had been planted. It was the first time we learned of tables to hold data in a database, which held data records broken up by "fields".
